## List Comprehensions

```python
squares = [n ** 2 for n in range(1, 6)]
print(squares)
```

```python
evens = [n for n in range(1, 21) if n % 2 == 0]
print(evens)
```

## Tuples – Defining a Tuple

```python
dimensions = (1920, 1080)
print(dimensions)
```

## Looping Through All Values in a Tuple

```python
dimensions = (1920, 1080)
for value in dimensions:
  print(value)
```

```python
fruits = ("apple", "banana", "cherry")
for index, fruit in enumerate(fruits):
  print(index, fruit)
```

## Dictionaries – Accessing Values in a Dictionary

```python
student = {"name": "Ali", "age": 20, "course": "AI Lab"}
print(student)
print(student["name"])
```

## Adding New Key-Value Pairs

```python
student = {"name": "Ali", "age": 20}
student["course"] = "AI Lab"
print(student)
```

## Modifying Values in a Dictionary

```python
student = {"name": "Ali", "age": 20, "course": "AI Lab"}
student["age"] = 21
print(student)
```

## Removing Key-Value Pairs (del and pop)

```python
student = {"name": "Ali", "age": 21, "course": "AI Lab", "semester": "CS-VB"}
del student["semester"]
print(student)
```

```python
removed_value = student.pop("age")
print(student)
print("Removed age:", removed_value)
```

## Using get() to Access Values

```python
student = {"name": "Ali", "course": "AI Lab"}
print(student.get("name"))
```

## Looping Through a Dictionary

```python
student = {"name": "Ali", "age": 21, "course": "AI Lab"}
for key, value in student.items():
  print(key, "->", value)
```

## A List of Dictionaries

```python
students = [
{"name": "Ali", "marks": 85},
{"name": "Sara", "marks": 92},
{"name": "Bilal", "marks": 76},
]
for student in students:
  print(student["name"], "scored", student["marks"])
```
